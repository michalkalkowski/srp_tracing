This supplementary material presents a convergence study for the shortest ray path (SRP) forward model used in weld map tomography. The key model parameters determining the accuracy are the size of the grid cell and the number of seeds along each edge of a grid cell. The grid cell size (sometimes referred to as an anisotropic sub-region in the literature) is linked to how the orientations vary across the weld and the geometry of the weld (chamfer). Typical scales relevant to practical variation or orientations are in the order of \SI{1-2}{\milli\metre} \cite{moysan2003, gueudre2009, fan2015}, as confirmed by published material examinations of such welds. The size of the grid cell also affects the representation of the weld geometry; larger grids may result in a 'stair-casing' effect, affecting the accuracy of the results, but this should be verified for each specific case. 

The number of seeds per grid cell edge is the main parameter used to verify the accuracy and convergence of the presented SRP model. The literature offers a few examples of applying shortest ray path (Dijkstra-like) solvers to determining propagation times in thick-section welds \cite{nowers}. The quoted references suggest covering the domain of interest with a set of either regularly or randomly distributed nodes, whose distribution is indirectly linked to the size of `anisotropic  subregions'. The model presented here seeds each grid cell with a number of nodes and allows only propagation paths within a single cell. That is, any node can only be linked to nodes surrounding the same grid cell - no longer jumps are allowed, as in \cite{nowers}. The angular resolution is achieved by allowing a considerable number of paths across each cell.

We present the convergence and the accuracy of the method on a simple numerical example. The test case is a V-weld (thickness: \SI{36.8}{\milli\metre}, weld cap: \SI{40}{\milli\metre}, root gap: \SI{1}{\milli\metre} with local orientations following an Ogilvy map ($T=2; n=1.3$, \cite{ogilvy}).The parent material Young's modulus was \SI{203.3}{\giga\pascal}, Poisson's ratio: 0.272, and density \SI{7900}{\kilogram\per\cubic\metre}. The transversely isotropic elastic constants for the weld material were: $C_{11}=C_{22}= 262$ \unit{\giga\pascal}, $C_{33}=229$ \unit{\giga\pascal}, $C_{12} = 148$ \unit{\giga\pascal}, $C_{13}=C_{23}=160$  \unit{\giga\pascal}, $C_{44}=C_{55}=82$ \unit{\giga\pascal}, $C_{66}=57$ \unit{\giga\pascal}, and the density \SI{8000}{\kg\per\cubic\metre}. Two 32-element arrays (pitch: \SI{2.05}{\milli\metre}, element width: \SI{1.55}{\milli\metre}) are placed in a through-transmission configuration - transmitting at the top and receiving at the bottom faces of the sample. We modelled propagation times across all transmitter-receiver combination pairs. The SRP model grid coincided with the Ogilvy grid and the only parameter to control the accuracy of the SRP solver was the number of seeds per edge of the grid. 

In addition, we executed an analogous time-domain finite element simulation as a benchmark. The FE model used the same Ogilvy map, with the same grid cell size. The central frequency was \SI{5}{\MHz} and the mesh size was \SI{0.025}{\mm} and the time step was just above \SI{4}{\ns}. Arrival times were extracted from FE-simulated traces based on the AIC criterion \cite{st-onge2011} with parameters tuned to propagation through an isotropic medium.

The reported target accuracy for time-of-flight modelling is \SI{0.05}{\micro\second} \cite{nowers}. In this study we look at the mean absolute error across all propagation paths and we consider 1 ans \SI{2}{\milli\metre} grids with increasing number of seeds per grid cell. \fref{fig:convergence_mean} show that the forward model converges and that the error is well within the target outlined in \cite{nowers}. A more detailed study of convergence and accuracy of the model is outside of the scope of this paper, but the presented results offer sufficient confidence for using the SRP model in the inversion scheme.

The method converges very quickly, however some residual error remains and cannot be resolved by increasing the number of seeds per edge. The reasons for the discrepancy are not a subject of concern since the error is insignificant for the application of interest. They may be related to the time-of-arrival extraction from the FE data, too. The two graphs above show both the mean and the maximum error. The maximum error is usually dominated by the receivers close to the weld root, so it is very localised. The mean error is usually well below 0.05 us which satisfies practical requirements. An illustrative error map for a configurations which can be defined as converged (8 seeds per edge) is shown below.
